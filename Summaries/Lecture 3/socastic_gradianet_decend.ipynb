### Overview
Stochastic gradient descent (SGD) is an iterative method to minimize an objective function expressed as an average over data points by using cheap, noisy estimates of the full gradient instead of computing the exact gradient on the entire dataset. SGD trades a higher-variance update at each step for much lower per-step cost, which makes it the workhorse optimizer for large-scale and deep learning problems.

### Objective, notation, and core formula
- Problem: minimize empirical risk  
  $$
  F(\theta)=\frac{1}{N}\sum_{i=1}^{N}\ell(\theta; x_i,y_i),
  $$
  where θ are parameters, $(x_i,y_i)$ are training examples, and ℓ is the per-example loss.  
- Exact gradient: $\nabla F(\theta)=\frac{1}{N}\sum_{i=1}^{N}\nabla_\theta \ell(\theta;x_i,y_i)$. Computing this every step is expensive for large N.  
- Stochastic (unbiased) gradient: pick a random index i (or a random mini-batch B) and use the estimate  
  $$
  g(\theta)=\nabla_\theta \ell(\theta;x_i,y_i)\quad\text{or}\quad g(\theta)=\frac{1}{|B|}\sum_{i\in B}\nabla_\theta \ell(\theta;x_i,y_i),
  $$
  which satisfies E[g(θ)] = ∇F(θ) under i.i.d. sampling assumptions. This unbiasedness is fundamental to the theoretical properties of SGD.

### Basic SGD algorithm and its intuition
- Update rule (single example):
  $$
  \theta_{t+1}=\theta_t-\eta_t\,\nabla_\theta\ell(\theta_t;x_{i_t},y_{i_t}),
  $$
  where η_t is the learning rate and i_t is a random sample drawn at step t.  
- Mini-batch variant (typical in practice):
  $$
  \theta_{t+1}=\theta_t-\eta_t\frac{1}{|B_t|}\sum_{i\in B_t}\nabla_\theta\ell(\theta_t;x_i,y_i).
  $$
- Intuition: each step moves parameters in a direction that would reduce loss for one (or a few) examples. Noise in the step can help escape shallow local minima and explore parameter space, while repeated sampling averages the noise and converges toward a minimizer in expectation.

### Mini-batch, batching tradeoffs, and common variants
- Full-batch gradient descent uses the exact gradient (costly per step but low variance); SGD uses single-example updates (cheap per step but high variance); mini-batches interpolate between them and are standard because they exploit parallel hardware while reducing gradient variance.  
- Common modern variants add mechanisms to stabilize and accelerate learning:  
  - Momentum: accumulates an exponential moving average of past gradients to damp oscillations and accelerate along consistent directions.  
  - RMSProp / AdaGrad: scale updates per-parameter using running estimates of squared gradients to adapt step sizes.  
  - Adam: combines momentum and adaptive per-parameter scaling, making learning-rate tuning easier in many problems. These variants are widely used because they improve convergence speed and robustness to hyperparameters.

### Step-size (learning rate), schedules, and convergence behavior
- Learning rate η_t is the most important hyperparameter. Fixed small η works in some settings; decreasing schedules (η_t → 0) are needed for classical convergence guarantees to stationary points. Practical schedules include step decay, exponential decay, and cosine annealing.  
- Theory (high level): with unbiased gradients and appropriate decreasing step sizes, SGD converges to stationary points for smooth objectives; for convex objectives, SGD converges to the global minimum in expectation at sublinear rates. For nonconvex deep networks, theory only guarantees convergence to stationary points under assumptions, but empirically SGD often finds good solutions.  
- Noise from stochastic gradients induces a “temperature” that can help escape saddle points and poor local minima; balancing noise (via batch size) and step size is a practical art.

### Practical recipe, diagnostics, and pitfalls
- Typical recipe for deep nets: use mini-batches (32–1024 depending on hardware), Adam or SGD+momentum, a starting learning rate (e.g., 1e-3 for Adam, 0.1–0.01 for SGD with momentum), and a schedule to reduce the rate when validation loss plateaus. Use weight decay, normalization (batchnorm/LayerNorm), and data augmentation to help generalization. These practices are empirical but effective in practice.  
- Diagnostics to watch: training loss, validation loss, gradient norms, distribution of activations, and whether parameters or gradients are saturating or vanishing. Instability signs: exploding gradients (reduce lr, gradient-clipping), vanishing gradients (architecture/activation choices), or very noisy/no-improving loss (reduce lr or change optimizer).  
- Pitfalls: using too large a learning rate causes divergence; too small makes progress painfully slow; overfitting with overly large models and small data; relying blindly on adaptive optimizers may produce worse generalization in some cases compared to well-tuned SGD+momentum.
